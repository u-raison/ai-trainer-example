```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import inspect

#解决乱码问题
plt.rcParams ['font.sans-serif'] = ['SimHei']
plt.rcParams ['axes.unicode_minus'] = False
```

```python
#定义条形图函数
def plot_bar_chart():

    """
    条形图 - 用于展示类别间的数值对比
    考试常见: 各科成绩对比、销售额对比等
    """

    #输出提示
    print("=" * 60)
    print("1. 绘制条形图")
    print("=" * 60)

    #创建一个字典类型数据
    data = {
    '水果': ['苹果', '香蕉', '樱桃', '葡萄'],
    '数量': [10, 15, 7, 10]
    }

    #将字典转换为二维表
    df = pd.DataFrame(data)

    #输出转换后的数据
    print("数据")
    print(df)

    #方法1：使用DataFrame.Plot() 推荐，简介
    print("\n方法1，df.plot()绘制")

    #绘制条形图，kind表示绘制图形类型，x表示x轴用水果名称这列，y表示y轴用数量这列，title表示图形标题，color表条形颜色，legend表示图例
    df.plot(kind='bar', x='水果',y='数量',color='skyblue',legend=False)

    #设置X、Y轴的标签
    plt.xlabel('水果类型')
    plt.ylabel('数量')
    
    #x标签不旋转
    plt.xticks(rotation = 0)

    #设置自动调整布局
    plt.tight_layout()

    #将当前图表保存为图片，dpi分辨率，bbox_inches自动裁剪多余空白
    plt.savefig ('水果销量条形图.png',dpi = 100,bbox_inches = 'tight')

    #输出提示
    print('条形图已保存：条形图')
    
    plt.show()

    #方法2： 使用plt.bar()更灵活
    print('\n方法2: plt.bar()绘制')

    #创建一个新的窗口,窗口高8，宽5
    plt.figure(figsize =(8,5))

    #绘制条形图，调出水果作为x轴，数量作为y轴，color设置颜色，width设置条形宽度
    plt.bar(df['水果'],df['数量'])

    #设置图形的标题并设置标题字体大小
    plt.title('水果数量分布',fontsize=14)

    #设置x轴和y轴的标签
    plt.xlabel('水果')
    plt.ylabel('数量')

    #设置风格线
    plt.grid(axis='y',alpha=0.3)

    plt.show()
```

```python
def plot_pie_chart():

    """
    饼图 - 用于展示各部分占总体的比例
    考试常见: 占比分析、百分比展示
    """

    #输出
    print("\n" + "=" * 60)
    print("2. 绘制饼图")
    print("=" * 60)

    #创建一个字典类型数据
    data = {
    '水果': ['苹果', '香蕉', '樱桃', '葡萄'],
    '数量': [10, 15, 7, 10]
    }

    #将字典转换为二维表
    df=pd.DataFrame(data)

    #输出提示
    print("数据：")
    print(df)

    print("\n绘制饼图...")

    #创建一个窗口
    plt.figure(figsize=(8,8))

    #绘制饼图，用数量作为扇形区大小
    plt.pie(
        df['数量'],
    
    #用水果这列作为扇形区的每块标签名
        labels = df['水果'],

    #显示百分比
    autopct = '%1.1f%%',

    #设置饼图从90度方向开始绘制
    startangle = 90,

    #设置每个扇形区的颜色
    colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99'],

    #设置扇形区是否突出显示
    explode = [0,0,0,0])

    #设置图形标题和标题字体大小
    plt.title('水果数量占比',fontsize=14)

    #设置饼图显示为正圆
    plt.axis('equal')

    #自动调整布局
    plt.tight_layout()

     #保存饼图图片
    plt.savefig("水果数量占比饼图.png",dpi=100,bbox_inches='tight')

    #显示图表
    plt.show()
```

```python
#定义绘制折线图的函数

def plot_line_chart():

    #输出提示
    print("\n" + "=" *60)
    print("3. 折线图绘制")
    print("=" * 60)

    #创建字典类型数据
    data = {
    '日期':['2025-01-01','2025-02-01','2025-03-01','2025-04-01'],
    '销售额':[200,220,240,260]
        }

    #数据类型转换，将字典类型数据转换为表格类型数据
    df = pd.DataFrame(data)

    #将日期这列数据转换为datatime日期时间类型
    df['日期'] = pd.to_datetime(df['日期'])

    #输出提示
    print("数据：")
    print(df)

    #方法1，使用DataFrame.plot()
    print("\n方法1：pd.plot()绘制")

    df.plot(
        kind='line',
        x='日期',#设置日期为X轴
        y='销售额',#设置销售额为Y轴
        title='月度销售额趋势', #添加折线图的标题
        marker='o', #添加数据标记点
        color ='green',#折线的颜色
        legend=False #不显示图例
        )

    #设置x、y轴的标签
    plt.xlabel ('日期')
    plt.ylabel ('销售额')

    #添加风格线，alpha设置网格线的透明度
    plt.grid(True,alpha=0.3)

    #自动布局
    plt.tight_layout()

    #将图形保存为图片，dpi分辨率，自动裁剪空白区域
    plt.savefig("月度销售趋势折线图.png",dpi =100,bbox_inches='tight')

    #输出提示
    print("折线图已保存:月度销售趋势折线图.png")

    plt.show()


    #方法2：使用plt.plot
    print("方法2：使用plt.plot()绘制")

    #创建新窗口
    plt.figure(figsize=(8,5))

    #绘制折线图
    plt.plot(
        df['日期'],df['销售额'], #日期作为折线图
        marker='o', #标记为原点
        markersize=8, #标记点大小8px
        linestyle='-',#折线为实线
        linewidth=2,#折线宽度
        color='red',#折线颜色
        label='销售额'#折线图例名称
    )

    #遍历销售额这一列，i索引位置，value对应当前销售额数值
    for i,value in enumerate(df['销售额']):
        print(i,value)
        #在图表中添加文字标签
        plt.text(df['日期'][i],
        value +5,
        str(value),
        ha='center', 
        va='bottom',
        fontsize= 15
        ) 

    #设置图形标题及标题字体大小
    plt.title("月度销售额趋势",fontsize=15)
    
    #设置x轴标签，y轴的标签
    plt.xlabel('日期')
    plt.ylabel('销售额（元）')

    #设置图例
    plt.legend()

    #设置网格线透明，线0.3px
    plt.grid (True,alpha=0.3)
    
    #设置x轴标签旋转15度
    plt.xticks(df['日期'], rotation=45)
    
    #自动布局
    plt.tight_layout()

    #显示图形
    plt.show()

```

```python
#定义绘制综合图表的函数
def comprehensive_example():

    """
    综合示例: 一次性绘制多个子图
    考试加分项: 展示更高级的可视化能力
    """

    #输出提示
    print("\n" + "="*60)
    print("4. 综合示例 - 多子图布局")
    print("="*60)

    #创建字典数据
    data = {
    '科目': ['语文', '数学', '英语', '物理'],
    '成绩': [85, 92, 78, 88]
    }

    #数据类型转换
    df = pd.DataFrame(data)

    #创建一个2行2列的4个子图的窗口
    fig,axes = plt.subplots(2,2,figsize=(12,10))

    #设置整张图的总标题
    fig.suptitle('学生成绩分析报告',fontsize=16,fontweight='bold')

    #绘制第1个子图，且是条形图
    axes[0,0].bar(df['科目'],df['成绩'],color='skyblue') #使用科目作为X轴，成绩作为Y轴，条块颜色天蓝色
    axes[0,0].set_title('各科成绩对比') #设置标题
    axes[0,0].set_ylabel('成绩') #设置y轴标签
    axes[0,0].grid(axis='y',alpha=0.3) #添加横向风格线

    #绘制第2个子图，且是饼图
    axes[0,1].pie(df['成绩'],labels=df['科目'],autopct='%1.1f%%',startangle=90,frame=True) #设置饼块数据，饼块标签，百分比保留2位，90度开始分割
    axes[0,1].set_title('成绩占比') #设置饼图标题
    #axes[0,1].set_frame_on(True)

    #绘制第3个子图，且是折线图
    axes[1,0].plot(df['科目'],df['成绩'],color='purple')#设置折线图X轴为科目，y轴为成绩，折线紫色
    axes[1,0].set_title('成绩占比') #设置折线图标题
    
    #绘制第4个子图，且是水平条形图
    axes[1,1].barh(df['科目'],df['成绩'],color='coral') #设置水平条形图，x轴为科目，成绩为y轴，珊瑚色
    axes[1,1].set_title('成绩排名') #设置折线图标题
    axes[1,1].set_xlabel('成绩') #设置x轴标签
    
    #自动调整布局
    plt.tight_layout()

    plt.savefig("综合分析图.png",dpi=100,bbox_inches='tight') #保存图片，设置分辨率100dpi，剔除多余空白

    #输出提示
    print("综合分析图片已保存：综合分析图.png")

    plt.show()


```

```python
#定义主函数
def main():
    print("数据可视化考点示例程序")
    print("=" * 60)

    #调用绘制3种基本图形的函数
    plot_bar_chart()
    plot_line_chart()
    plot_pie_chart()

    #调用绘制综合图形的函数
    comprehensive_example()

    print("\n" + "=" * 60)
    print("所有图表绘制完成！")
    print("=" * 60)
```

```python
#python程序主入口
if __name__ == "__main__":
    main()
```
