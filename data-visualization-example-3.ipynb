```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']
```

```python
file_path = "fruit.csv"
df = pd.read_csv(file_path)

print("数据")
print(df)
```

```python
#绘制条形图
plt.figure(figsize=(8,5))
plt.bar(
    df['水果'],
    df['数量'],
    color = ['#FF7F50', '#FFD700', '#FFA500', '#FF6347']
    )

plt.title('不同水果的数量对比',fontsize=12)
plt.xlabel('水果类型',fontsize=10)
plt.ylabel('数量',fontsize=10)

for i,value in enumerate(df['数量']):
    #df['数量'][i]
    plt.text(i, value+5,str(value),ha='center')

plt.tight_layout()

plt.show()
```

```python

plt.figure(figsize=(8,8))
plt.pie(
    df['数量'],
    labels=df['水果'],
    autopct='%1.1f%%',
    colors=['#FF7F50', '#FFD700', '#FFA500', '#FF6347'],
    startangle=90
)

plt.title("各类水果数量占比",fontsize=12)
plt.axis('equal')
plt.tight_layout
plt.show()

```

```python
#绘制折线图

df_line = pd.DataFrame({
    '季度': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Apple销量': [110, 120, 100, 130],
    'Banana销量': [170, 180, 160, 190]
})

plt.figure(figsize=(8,5))

#绘制apple的销量折线
plt.plot(

    df_line['季度'],
    df_line['Apple销量'],
    marker ='o',
    label = 'Apple',
    color = '#FF7F50'
)

#绘制banner的销量折线
plt.plot(

    df_line['季度'],
    df_line['Banana销量'],
    marker = 's',
    label ='Bananer',
    color = '#FFD700'
)

plt.title('水果季度销量趋势图',fontsize=12)
plt.xlabel('季度',fontsize=10)
plt.ylabel('销量',fontsize=10)

plt.legend()
plt.tight_layout()
plt.show()
```
